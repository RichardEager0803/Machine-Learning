**How did you identify and handle implicit missing values (such as -200 sentinel indicators) across the sensor columns, and how did your imputation strategy affect model stability?**

* The UCI dataset uses an implicit missing value of -200 to represent sensor errors or missing telemetry instead of standard NaN values. I was able to catch this in my program during ingestion using na_values=[-200].

* I used SimpleImputer(strategy="median") for numerical features, and most_frequent for categoric ones. Median is not as sensitive to outliers like the mean, which can get easily skewed.

* When filling the gaps rather than dropping entire rows of sensor data, the training set could retain a large sample size. This helps to prevent information loss. It also ensures that the model properly converges and allows us to avoid runtime errors during matrix operations.

**Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps in your cloud pipeline.**

* The pipeline lets us chain transformers together. For example, we can do imputation and scaling in a single object. While the ColumnTransformer applies different steps to specific feature types all at the same time. For example, the StandardScaler on numeric data and OneHotEncoder on categorical data. These allow us to eliminate some of the manual steps and also lets us maintain the exact matrix alignment.

* The pipeline prevents data leakage by computing the parameters purely on the training data set during .fit() and then applies those exact metrics to test sets during the .transform() so that the data is isolated.

**How did comparing StandardScaler versus MinMaxScaler impact your model training convergence, gradient descent optimization, and final evaluation metrics?**

* StandardScaler standardizes features to have a mean of 0 and variance of 1, while the MinMaxScaler bounds features between 0 and 1.

* The air quality dataset can contain some extreme peaks (spikes in chemical pollutants). So MinMaxScaler could be sensitive to these outliers since a single extreme maximum pollutant reading can squash all the other normal readings into a small range close to 0.

* Standard Scaler handles the outliers much better, which leads to smoother gradient descent optimization, and faster convergence.

**How does configuring SGDClassifier differ from standard closed-form solvers, and what advantages do probability outputs and evaluation metrics provide when assessing classification performance?**

* Standard Logistic regression uses batch optimization to solve the function globally. While SGDClassifier uses stochastic gradient descent, it updates the weights iteratively using small data batches or single samples. This makes it much more scalable for larger datasets.

* Log loss allows the model to output continuous probabilities as opposed to discrete class labels. This lets us use advanced evaluation metrics such as ROC-AUC.

**Based on your regularized regression models (Ridge versus Lasso), how did adjusting the regularization hyperparameter affect feature selection and coefficient shrinkage, and what operational environmental insights do these findings reveal?**

* The higher the regularization parameter, the more that larger model weights will be penalized.

* For Ridge (L2), the coefficients are smoothly shrunk towards zero without actually reaching zero. While Lasso (L1), can penalize the coefficients to zero, which effectively performs automated feature selection.

* Lasso reveals which chemical sensors and meteorological factors such as metal oxide or humidity, are truly drivers of carbon monoxide levels. This can help environmental agencies streamline physical sensor arrays, by removing redundant or uninformative telemetry lines.
