1. Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.

>* The missing values in this dataset for diabetes use 0 as a placeholder. For each of these physiological features (besides the target and pregnancies), the placeholder value needs to be imputed or else there's risk in skewing the data distribution. 3. What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?

>* Out of bag (OOB) training allows us to estimate model performance using the training observations that were not selected in the bootstrap sample for a particular decision tree. Since each tree is trained using a random sample, with replacement, some data is not used from the training sample. This omitted data can be used to evaluate the tree. Their predictions are then aggregated across the relavent trees in order to calculate the OOB score.

>* OOB evaluation provides an estimate of generalization performance without requiring cross-validation. This helps us save computational time since the bootstrap samples that are used during the Bagging create the validation observations. For my implementation, the oob_score=True calculates the OOB score, which I compare with accuracy on the held-out test set. Similar scores can suggest that the estimates are reasonably consistent, while differences can suggest sampling variability or differences between the evaluation procedures. OOB evaluation can be efficient but it doesn't outright replace cross validation in every situation.


4. How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?

>* Standard Bagging trains multiple models on different samples of the training data, but each of the decision trees can consider all available features when selecting a split. The random forests add randomness by restricting the features considered at each split. In my implementation, max_features="sqrt" lets each split consider roughly the square root of the total number of features.

>* The randomness reduces the correlation between individual trees since they are less likely to make identical splitting decisions. When the predictions of each of the trees are aggregated, lower correlation can reduce the variance of the ensemble and improve generalization on new data. Though restricting the features can make a single, individual tree less accurate, so there's a tradeoff between the tree's strenght and its diversity. Random Forests usually generalize better than Bagging when feature subsetting reduces the correlation between the trees.
