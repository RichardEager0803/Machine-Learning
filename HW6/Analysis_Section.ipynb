3. Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?

>* The PCA must be fitted strictly on the training data rather than the entire set to prevent data leakage. Fitting on the whole set would allow test images to influence the calculated mean and principal components, even though their labels are not used.

>* This can make the evaluation less representative of real world performance since information from the test set has influenced preprocessing.

>* The proper approach would be to fit the PCA on the training data and use the same transformation on the test data.
