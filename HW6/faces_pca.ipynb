# **Assignment #6**

Dimensionality Reduction, Reconstruction, and
Non-Linear Classification on Face Images

**Goals:**
* Examine Principal Component Analysis (PCA) and dimensionality reduction using the classic **Olivetti Faces dataset**
* Inspect high-dimensional image vectors, build Scikit-Learn pipelines to analyze explained variance, reconstruct compressed images, and visualize lower dimensional feature spaces
* Systematically evaluate how PCA compression impacts downstream classification performance across linear, non-linear (Kernel SVM), and ensemble classifiers


**Dataset Overview:**

The Olivetti Faces dataset contains 400 grayscale images (64 x 64 pixels) collected across 40 distinct individuals (10 images per subject) taken under varying lighting conditions, facial expressions, and facial details (e.g., glasses / no glasses).
* The **target variable (target)** is an integer label ranging from 0 to 39 representing the subject identity
* The **image features (data)** consist of a set of flattened 4,096-dimensional feature vector (64 x 64 = 4,096) with floating-point pixel intensities normalized between 0.0 and 1.0
* The **image matrices (images)** consist of the raw 64 x 64 2D representations used for image rendering and reconstruction visualization


**Imports:**

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    mean_squared_error
)

from sklearn.datasets import fetch_olivetti_faces
```

Task 1 - Dataset Loading and Dimensionality Inspection
* Loads the Olivetti Faces dataset, and prints the total number of samples, unique target subjects, feature dimensions, and plots a 4x5 grid of sample faces labeled with their subject IDs.

```python
# loads the dataset
faces = fetch_olivetti_faces()

# Extracts data features, target label, and images
X = faces.data
y = faces.target
images = faces.images

# Prints the dataset information
print("Total number of samples:", X.shape[0])
print("Number of unique target subjects:", len(set(y)))
print("Feature dimensions:", X.shape[1])
print("Image dimensions:", images.shape[1:])

# Plots a 4x5 grid of sample faces
fig, axes = plt.subplots(4, 5, figsize=(12, 10))

for i, ax in enumerate(axes.flat):
  ax.imshow(images[i], cmap="gray")
  ax.set_title(f"Subject ID: {y[i]}")
  ax.axis("off")

plt.tight_layout()
plt.show()
```

Task 2 - Cumulative Explained Variance & Elbow

* Plot Fits a full PCA model on the dataset. Plots both the individual explained variance ratio and the cumulative explained variance curve versus the number of principal components.
* Determines and prints the exact number of components required to preserve 90%, 95%, and 99% of the cumulative variance.

```python
# Fits PCA using all available components
pca = PCA()
pca.fit(X)

# Calculates individual and cumulative explained variance
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

components = np.arange(1, len(explained_variance) + 1)

print("Number of PCA components:", pca.n_components_)

for threshold in [0.90, 0.95, 0.99]:
  n_components = np.argmax(cumulative_variance >= threshold) + 1
  actual_variance = cumulative_variance[n_components - 1]

  print(f"{threshold * 100:.0f}% variance")
  print(f"{n_components} components")
  print(f"(cumulative variance = {actual_variance:4%})")

# Plots individual and cumulative explained variance
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Individual explained variance
axes[0].bar(components, explained_variance, color="blue")
axes[0].set_title("Individual Explained Variance")
axes[0].set_xlabel("Principal Component")
axes[0].set_ylabel("Explained Variance Ratio")
# axes[0].set_yscale("log")
axes[0].grid(axis="y", linestyle="--", alpha=0.4)

# Cumulative explained variance
axes[1].plot(
  components,
  cumulative_variance,
  color="red",
  linewidth=2
)

colors = ["orange", "green", "purple"]

for threshold, color in zip([0.90, 0.95, 0.99], colors):
  n = np.argmax(cumulative_variance >= threshold) + 1

  axes[1].axhline(
    y=threshold,
    color=color,
    linestyle="--",
    label=f"{threshold:.0%} variance"
  )
  axes[1].axvline(
    x=n,
    color=color,
    linestyle=":",
    alpha=0.7
  )
  axes[1].scatter(
    n, cumulative_variance[n - 1],
    color=color, zorder=3
  )

axes[1].set_title("Cumulative Explained Variance")
axes[1].set_xlabel("Number of Principal Components")
axes[1].set_ylabel("Cumulative Explained Variance")
axes[1].set_ylim(0, 1.02)
axes[1].legend()
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

Task 3 - Image Compression and Reconstruction Analysis

* Selects 5 representative facial images. Compresses and reconstructs them using PCA with varying numbers of components; e.g., k in {10, 50, 100, 200, 300}.
* Plots a grid comparing the original images against their reconstructions across each component threshold.
 * Computes and plots the Mean Squared Error (MSE) between original and reconstructed faces as a function of k.

```python
# Number of principal components to evaluate
k_values = [10, 50, 100, 200, 300]

# Selects 5 representative faces from different subjects
sample_indices = [0, 50, 100, 150, 200]
original_faces = X[sample_indices]

# Fits PCA using the maximum number of components needed
pca_reconstruction = PCA(n_components=max(k_values))
pca_reconstruction.fit(X)

# Stores reconstructed faces and MSE values
reconstructions = {}
mse_values = []

for k in k_values:
  # Compresses and reconstructs the selected faces
  pca_k = PCA(n_components=k)
  pca_k.fit(X)

  compressed = pca_k.transform(original_faces)
  reconstructed = pca_k.inverse_transform(compressed)

  reconstructions[k] = reconstructed

  # Calculates MSE across all 5 faces and all pixels
  mse = mean_squared_error(
    original_faces.ravel(),
    reconstructed.ravel()
  )
  mse_values.append(mse)

# Plots original images and reconstructions
fig, axes = plt.subplots(
  len(sample_indices),
  len(k_values) + 1,
  figsize=(15, 12)
)

for row, idx in enumerate(sample_indices):
  # Original face
  axes[row, 0].imshow(faces.images[idx], cmap="gray")
  axes[row, 0].set_title("Original" if row == 0 else f"Subject {faces.target[idx]}")
  axes[row, 0].axis("off")

  # Reconstructed faces for each k
  for col, k in enumerate(k_values, start=1):
    reconstructed_face = reconstructions[k][row].reshape(64, 64)

    axes[row, col].imshow(
      reconstructed_face,
      cmap="gray",
      vmin=0,
      vmax=1
    )

    if row == 0:
      axes[row, col].set_title(f"k = {k}")

    axes[row, col].axis("off")

plt.suptitle("Original vs. PCA-Reconstructed Faces", fontsize=16)
plt.tight_layout()
plt.show()

print("Mean Squared Error (MSE) by number of components:")
for k, mse in zip(k_values, mse_values):
  print(f"k = {k:3d}: MSE = {mse:.6f}")

# Plots MSE as a function of k
plt.figure(figsize=(8, 5))
plt.plot(k_values, mse_values, marker="o", linewidth=2, color="orange")
plt.xlabel("Number of Principal Components (k)")
plt.ylabel("Mean Squared Error (MSE)")
plt.title("PCA Reconstruction Error vs. Number of Components")
plt.xticks(k_values)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

Task 4 - Visualization of "Eigenfaces"
* Extracts the principal components (eigenvectors) from the fitted PCA model.
* Reshapes the first 15 principal components back into 64 x 64 grids and puts them in a 3 x 5 plot grid ("Eigenfaces").
* Each plot is annotated with the percentage of total variance explained by that specific component.

```python
# Extracts the first 15 principal components
eigenfaces = pca.components_[:15]

# Individual explained variance for each component
eigenface_variance = pca.explained_variance_ratio_[:15] * 100
print("Eigenfaces shape:", eigenfaces.shape)

# Plots the first 15 eigenfaces in a 3 x 5 grid
fig, axes = plt.subplots(3, 5, figsize=(14, 9))

for i, ax in enumerate(axes.flat):
  eigenface = eigenfaces[i].reshape(64, 64)

  ax.imshow(eigenface, cmap="gray")
  ax.set_title(
    f"Eigenface {i + 1}\n"
    f"Variance: {eigenface_variance[i]:.3f}%"
  )
  ax.axis("off")

fig.suptitle("First 15 Eigenfaces", fontsize=16)
plt.tight_layout()
plt.show()
```

Task 5 - Classification Baseline - Raw vs PCA-Reduced Features (Linear Model)
* Performs a stratified train/test split (80/20).

* Trains a LogisticRegression classifier on:

>1. The raw 4,096-dimensional feature vectors.

>2. The PCA-reduced feature vectors using the 95% variance threshold fitted strictly on the training set.

```python
X_train, X_test, y_train, y_test = train_test_split(
  X,
  faces.target,
  test_size=0.20,
  stratify=faces.target,
  random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# Fits PCA ONLY on the training data.
# Retains enough components to explain at least 95% of variance.
pca = PCA(n_components=0.95, svd_solver="full")
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print("PCA components required for 95% variance:", pca.n_components_)
print("Training variance retained:", f"{np.sum(pca.explained_variance_ratio_):.4%}")
print("Reduced feature dimensions:", X_train_pca.shape[1])

results = {}
predictions = {}


# Logistic Regression on raw features
raw_logreg = make_pipeline(
  StandardScaler(),
  LogisticRegression(max_iter=3000, random_state=42)
)

start = time.perf_counter()
raw_logreg.fit(X_train, y_train)
raw_time = time.perf_counter() - start

raw_pred = raw_logreg.predict(X_test)
raw_accuracy = accuracy_score(y_test, raw_pred)

results["Logistic Regression\n(Raw)"] = {
    "Accuracy": raw_accuracy,
    "Training Time": raw_time
}
predictions["Logistic Regression (Raw)"] = raw_pred


# Logistic Regression on PCA-related features
pca_logreg = make_pipeline(
  StandardScaler(),
  LogisticRegression(max_iter=3000, random_state=42)
)

start = time.perf_counter()
pca_logreg.fit(X_train_pca, y_train)
pca_logreg_time = time.perf_counter() - start

pca_logreg_pred = pca_logreg.predict(X_test_pca)
pca_logreg_accuracy = accuracy_score(y_test, pca_logreg_pred)

results["Logistic Regression\n(PCA)"] = {
    "Accuracy": pca_logreg_accuracy,
    "Training Time": pca_logreg_time
}
predictions["Logistic Regression (PCA)"] = pca_logreg_pred

print("\nLogistic Regression comparison")
print(f"Raw features: {raw_accuracy:.4%} accuracy, {raw_time:.4f} seconds")
print(f"PCA features: {pca_logreg_accuracy:.4%} accuracy, "f"{pca_logreg_time:.4f} seconds")


# Dual-axis plot for the two Logistic Regression Models
baseline_names = [
  "Logistic Regression\n(Raw)",
  "Logistic Regression\n(PCA)"
]

baseline_accuracy = [
  results[name]["Accuracy"] for name in baseline_names
]
baseline_time = [
  results[name]["Training Time"] for name in baseline_names
]

fig, ax1 = plt.subplots(figsize=(9, 5))
ax2 = ax1.twinx()

positions = np.arange(len(baseline_names))
width = 0.35

bars1 = ax1.bar(
  positions - width / 2,
  baseline_accuracy,
  width,
  color="blue",
  label="Test Accuracy"
)
bars2 = ax2.bar(
  positions + width / 2,
  baseline_time,
  width,
  color="red",
  label="Training Time"
)

ax1.set_ylabel("Test Accuracy")
ax2.set_ylabel("Training Time (seconds)")
ax1.set_xticks(positions)
ax1.set_xticklabels(baseline_names)
ax1.set_ylim(0, 1.05)
ax1.set_title("Raw vs. PCA Logistic Regression")
ax1.bar_label(bars1, fmt="%.3f", padding=3)
ax2.bar_label(bars2, fmt="%.3f s", padding=3)

handles = [bars1, bars2]
labels = [h.get_label() for h in handles]
ax1.legend(
    handles,
    labels,
    loc="upper left",
    bbox_to_anchor=(1.08, 1),
    frameon=True,
)
ax1.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Task 6 - Non-Linear Classification (Kernel SVM)
* Trains an SVC with an RBF kernel on the PCA-reduced features (95% variance).

```python
rbf_svm = make_pipeline(
  StandardScaler(),
  SVC(kernel="rbf", probability=True, random_state=42)
)

start = time.perf_counter()
rbf_svm.fit(X_train_pca, y_train)
rbf_time = time.perf_counter() - start

rbf_pred = rbf_svm.predict(X_test_pca)
rbf_accuracy = accuracy_score(y_test, rbf_pred)

results["RBF SVM\n(PCA)"] = {
  "Accuracy": rbf_accuracy,
  "Training Time": rbf_time
}
predictions["RBF SVM (PCA)"] = rbf_pred

print("\nRBF SVM")
print(f"Test accuracy: {rbf_accuracy:.4%}")
print(f"Training time: {rbf_time:.4f} seconds")
```

Task 7 - Ensemble Classification on PCA Features
* Constructs a VotingClassifier (soft voting) combining a Polynomial Kernel SVM (degree=2), an RBF Kernel SVM, and a RandomForestClassifier (n_estimators=100).
* Trains the ensemble on the PCA-reduced features.

```python
poly_svm = SVC(
  kernel="poly",
  degree=2,
  probability=True,
  random_state=42
)

rbf_svm_ensemble = SVC(
  kernel="rbf",
  probability=True,
  random_state=42
)

random_forest = RandomForestClassifier(
  n_estimators=100,
  random_state=42,
  n_jobs=-1
)

voting = VotingClassifier(
  estimators=[
    ("poly_svm", poly_svm),
    ("rbf_svm", rbf_svm_ensemble),
    ("random_forest", random_forest)
  ],
  voting="soft",
  n_jobs=-1
)

ensemble = make_pipeline(
  StandardScaler(),
  voting
)

start = time.perf_counter()
ensemble.fit(X_train_pca, y_train)
ensemble_time = time.perf_counter() - start

ensemble_pred = ensemble.predict(X_test_pca)
ensemble_accuracy = accuracy_score(y_test, ensemble_pred)

results["Soft Voting Ensemble\n(PCA)"] = {
  "Accuracy": ensemble_accuracy,
  "Training Time": ensemble_time
}

predictions["Soft Voting Ensemble (PCA)"] = ensemble_pred

print("\nSoft Voting Ensemble")
print(f"Test accuracy: {ensemble_accuracy:.4%}")
print(f"Training time: {ensemble_time:.4f} seconds")
```

Classification summary (Tasks 6 & 7), comparing all models using the same test set

```python
summary = []

for model_name, y_pred in predictions.items():
  summary.append({
    "Model": model_name,
    "Accuracy": accuracy_score(y_test, y_pred),
    "Weighted Precision": precision_score(
      y_test, y_pred, average="weighted", zero_division=0
    ),
    "Weighted Recall": recall_score(
      y_test, y_pred, average="weighted", zero_division=0
    ),
    "Weighted F1": f1_score(
      y_test, y_pred, average="weighted", zero_division=0
    )
  })

summary_df = pd.DataFrame(summary)

print("\nCLASSIFICATION PERFORMANCE SUMMARY")
print(summary_df.round(4).to_string(index=False))

print("\nRBF SVM CLASSIFICATION REPORT")
print(classification_report(y_test, rbf_pred, zero_division=0))
```
