3. Compare the clustering behavior and cluster assignments of K-Means against DBSCAN and HDBSCAN on the 13-dimensional Wine feature space. Why did DBSCAN and HDBSCAN classify specific wine samples as noise, while K-Means assigned every sample to a cluster?

>* DBSCAN and HDBSCAN can classify certain wine samples as noise because of the fact that they are density based clustering algorithms and can thus label some samples as noise rather than belonging to a specific cluster.

>* K-Means on the other hand, assigns each sample the cluster which it is closest to, leaving no outliers behind. This is one of the flaws of K-Means in comparison to Density Based clustering algorithms.

4. Contrast the operational mechanisms of standard DBSCAN and HDBSCAN. How does HDBSCAN's hierarchical, stability-based approach overcome DBSCAN's vulnerability to a single global eps parameter when clustering chemical datasets with varying local densities?

>* HDBSCAN overcomes DBSCANS limitations by creating a hierarchical representation of clusters, across varying density levels. Instead of using one global epsilon value, it evaluates cluster stability using different density thresholds. The wine clusters which persist across a wider range of density levels are considered more stable and are selected as meaningful clusters. Since this dataset has varying local densities, HDBSCAN can use this to its advantage when clustering the data.
