4. Discuss how tuning hyperparameters such as regularization parameter C and kernel coefficient gamma in an SVC impacts the trade-off between bias, variance, and model generalization on unseen medical test data.

>* Tuning hyperparameters such as the regularization parameter C and the kernel coefficient gamma finds the balance between bias and variance by controlling how aggressively the model fits. A high C penalizes misclassifications whereas high gamma restricts each support vector's radius of influence, lowering bias but increasing variance. This leads to jagged decision boundaries that can overfit training noise, and then fail on new external patient data. Although extremely low gamma values will increase bias and underfitting, which will not generalize well on the data.
