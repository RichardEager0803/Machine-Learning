3.   Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing steps.

>*   Pipelines allow us to chain the transforms, e.g., imputation and scaling, into one single object. While the ColumnTransformer applies different steps to specific feature types all at the same time. For example, the StandardScaler on numeric data and OneHotEncoder for categorical data. This lets us eliminate some of the manual steps and also maintains exact matrix alignment.
>*   Data leakage occurs when the test data set statistics, e.g., mean or variance, leaks into the training. The Pipelines prevent this by computing the parameters purely on the training set during .fit() and then applying those exact metrics to test sets during the .transform() so the data is isolated.
